# Chargement des dépendances

```python
from transformers import AutoModelForCausalLM
from transformers import AutoTokenizer
from transformers import pipeline
import torch
```

# Chargement du modèle

```python
import gc

# Clean VRAM/RAM before (re)loading the model
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    try:
        torch.cuda.ipc_collect()
    except Exception:
        pass

# Modèle instruct "small" qui tient sur T4 16GB sans bitsandbytes
# (fp16 + device_map=auto)
model_id = "Qwen/Qwen2.5-3B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id, use_fast=True)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    torch_dtype=torch.float16,
 )

# (optionnel) évite certains warnings de padding selon les versions
if tokenizer.pad_token_id is None and tokenizer.eos_token_id is not None:
    tokenizer.pad_token_id = tokenizer.eos_token_id
```

# Prompt Zero Shot

```python
question = "Qu'est-ce que l'apprentissage supervisé ?"

# Construire une requête "chat" simple (format texte)
prompt = f"User: {question} Assistant: "

inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

generated_ids = model.generate(
    **inputs,
    max_new_tokens=500,
    do_sample=True,
    temperature=0.7,
    top_p=0.9,
    pad_token_id=tokenizer.pad_token_id,
)

# IMPORTANT: `generated_ids` contient [prompt + continuation].
# Si tu décodes tout, tu reverras le prompt. On découpe donc la partie "nouvelle".
prompt_len = inputs["input_ids"].shape[-1]
new_tokens = generated_ids[0, prompt_len:]
answer = tokenizer.decode(new_tokens, skip_special_tokens=True).strip()

print(answer)
```

On remarque que sans lui indiquer de limite de mots dans le prompt il est très verbeux et se confronte à sa limite de new tokens.

# Test de différents paramètres : 

```python
question = "Qu'est-ce que l'apprentissage supervisé ?"

# Construire une requête "chat" simple (format texte)
prompt = f"""
User: {dataset[0]['question']}
Assistant: {dataset[0]['answer']}
User: {question} Assistant: 
"""

# Valeurs intéressantes à tester
# - top_k=0 : désactive top-k (on se repose sur top_p si activé)
# - top_p=1.0 : désactive top-p (on se repose sur top_k si activé)
# - temperature basse : réponses plus "prudentes" ; plus haute : plus créatives
top_k_values = [0, 10, 20, 50, 100, 200]
top_p_values = [0.6, 0.75, 0.9, 0.95, 0.98, 1.0]
temperature_values = [0.2, 0.4, 0.7, 1.0, 1.3, 1.6]
answers_list = []

inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

print("PROMPT: \n ", prompt)

for i in range (len(top_k_values)):
    generated_ids = model.generate(
        **inputs,
        max_new_tokens=500,
        do_sample=True,
        temperature=temperature_values[i],
        top_p=top_p_values[i],
        top_k=top_k_values[i],
        pad_token_id=tokenizer.pad_token_id,
    )

    # IMPORTANT: `generated_ids` contient [prompt + continuation].
    # Si tu décodes tout, tu reverras le prompt. On découpe donc la partie "nouvelle".
    prompt_len = inputs["input_ids"].shape[-1]
    new_tokens = generated_ids[0, prompt_len:]
    answer = tokenizer.decode(new_tokens, skip_special_tokens=True).strip()
    answers_list.append(answer)

    print("-"*100)
    print(f"temperature = {temperature_values[i]}, top_p = {top_p_values[i]}, top_k = {top_k_values[i]}\n")
    print("Réponse du LLM : ",answer)
```

On voit que les réponses sont satisfaisantes dans l'ensemble, le modèles devient de plus en plus créatif avec les paramètres qu'on lui donne, sauf pour la dernière réponse où il l'est trop au point de choisir des tokens qui ne font aucun sens. Cela peut lui arriver aussi de générer une seconde question et d'y répondre.

```python
import re


def compare_answers(answer, ref_answer, parameters):
    # Longueurs (optionnel, pour analyse)
    len_answer = len(answer)
    len_ref_answer = len(ref_answer)


    # Tokenisation simple en mots (minuscules, Unicode)
    def tokenize(text: str) -> list[str]:
        tokens = re.findall(r"\b[\w']+\b", text.lower(), flags=re.UNICODE)
        return [t for t in tokens if len(t) > 1]


    words_answer = set(tokenize(answer))
    words_ref = set(tokenize(ref_answer))


    # Liste des mots en commun
    common_set = words_answer.intersection(words_ref)
    common_words = sorted(common_set)


    # Similarité / distance de Jaccard sur les mots uniques
    union_set = words_answer.union(words_ref)
    jaccard_similarity = (len(common_set) / len(union_set)) if len(union_set) else 0.0
    jaccard_distance = 1.0 - jaccard_similarity


    print("len(answer)=", len_answer, "len(ref_answer)=", len_ref_answer)
    print("Nombre de mots en commun :", len(common_words))
    print(f"Jaccard similarity = {jaccard_similarity:.3f} | Jaccard distance = {jaccard_distance:.3f}")
    return {
        "top_k": parameters[0],
        "top_p": parameters[1],
        "temperature": parameters[2],
        "len_answer": len_answer,
        "len_ref_answer": len_ref_answer,
        "num_common_words": len(common_words),
        "jaccard_similarity": jaccard_similarity,
        "jaccard_distance": jaccard_distance,
    }


ref_answer = dataset[4]["answer"]
metrics_list = []
for i in range(len(answers_list)):
    metrics = compare_answers(answers_list[i], ref_answer, [top_k_values[i], top_p_values[i], temperature_values[i]])
    print(metrics)
    metrics_list.append(metrics)
```

```python
import pandas as pd


# Conversion de la liste de métriques en tableau (DataFrame)
df_metrics = pd.DataFrame(metrics_list)


# Affichage du tableau dans le notebook
df_metrics
```

On remarque (du moins avec l'execution actuelle) que nous avons étonnament peu de mots en commun,  et que ce sont souvent les mêmes qui reviennent. On remarque aussi que les réponses du LLM sont bien plus longues que celle de référence. Il est intéressant de voir la métrique de la similarité de Jaccard (% d'éléments en commun aux deux réponses), on voit en effet que plus on augmente l'aléatoire du modèle, plus cette similarité semble décroître, ce qui est logique car on s'éloigne de plus en plus de la réponse attendue.

```python
# Expérience: question "Qu'est-ce qu'un embedding ?" + sweep paramètres
# One-shot: on fournit un exemple Q/R (comme en Cellule 18), puis on pose la question cible.

question = dataset[11]["question"]  # "Qu'est-ce qu'un embedding ?"
ref_answer = dataset[11]["answer"]

prompt = f"""
User: {dataset[9]['question']}
Assistant: {dataset[9]['answer']}
User: {question} Assistant: 
"""

answers_list_embedding = []
metrics_list_embedding = []

inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
print("PROMPT:\n", prompt)

for i in range(len(top_k_values)):
    generated_ids = model.generate(
        **inputs,
        max_new_tokens=500,
        do_sample=True,
        temperature=temperature_values[i],
        top_p=top_p_values[i],
        top_k=top_k_values[i],
        pad_token_id=tokenizer.pad_token_id,
    )

    # IMPORTANT: `generated_ids` contient [prompt + continuation].
    # On découpe donc la partie "nouvelle".
    prompt_len = inputs["input_ids"].shape[-1]
    new_tokens = generated_ids[0, prompt_len:]
    answer_i = tokenizer.decode(new_tokens, skip_special_tokens=True).strip()
    answers_list_embedding.append(answer_i)

    print("-" * 100)
    print(f"temperature = {temperature_values[i]}, top_p = {top_p_values[i]}, top_k = {top_k_values[i]}\n")
    print("Réponse du LLM : ", answer_i)

    metrics = compare_answers(
        answer_i,
        ref_answer,
        [top_k_values[i], top_p_values[i], temperature_values[i]],
    )
    metrics_list_embedding.append(metrics)

import pandas as pd

df_metrics_embedding = pd.DataFrame(metrics_list_embedding)
df_metrics_embedding
```
